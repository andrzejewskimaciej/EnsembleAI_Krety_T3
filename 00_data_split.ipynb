```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data/data.csv')
df.shape
```

W naszych danych tylko dane oznaczone jako `train` mają podaną wartość x2 (którą chcemy przewidywać). Dlatego też wydzielimy z tych danych wewnątrz dane treningowe i testowe. Przewidywać będziemy x2 w ostatnim miesiącu.

```python
df = df.loc[df['period'] == 'train']
df.shape
```

```python
df.head()
```

```python
df.timedate = pd.to_datetime(df.timedate)
df_test = df.loc[df['timedate'] >= '2025-04-01']
df_train = df.loc[df['timedate'] < '2025-04-01']
df_test.shape, df_train.shape
```

```python
df_test.to_csv('data/data_test.csv', index=False)
df_train.to_csv('data/data_train.csv', index=False)
```
